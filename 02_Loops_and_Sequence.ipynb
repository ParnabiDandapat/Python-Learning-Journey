```python
#To remove an element at a specific index in the list, you can use the pop() method like this:

numbers = [1, 2, 3, 4, 5]
numbers.pop(1) # The number 2 is returned
```

```python
#If you don't specify an index for the pop method, then the last element is removed.

numbers = [1, 2, 3, 4, 5]
numbers.pop() # The number 5 is returned
```

```python
words=['tree','mountain','sky','cloud','river']

def is_long_word(word):
  return len(word)>4

long_words=list(filter(is_long_word,words))
print(long_words)
```

```python
celsius = [0, 10, 20, 30, 40]

def to_fahrenheit(temp):
    return (temp * 9/5) + 32

fahrenheit = list(map(to_fahrenheit, celsius))
print(fahrenheit)
```
